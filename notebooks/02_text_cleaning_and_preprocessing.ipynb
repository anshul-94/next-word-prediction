```python
from pathlib import Path
# Project root directory
PROJECT_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()

# Original dataset path
RAW_DATA_PATH = PROJECT_DIR / "data" / "training_text.txt"

# Load original text
with open(RAW_DATA_PATH, "r", encoding="utf-8") as file:
    raw_text = file.read()

print("Original dataset loaded successfully!")
print("Dataset path:", RAW_DATA_PATH)
print("Data type:", type(raw_text))
print("Total characters:", len(raw_text))

print("\nFirst 500 characters:")
print(repr(raw_text[:500]))
```

```python

import unicodedata
from collections import Counter

# Count all characters
character_counts = Counter(raw_text)

# Inspect non-ASCII characters
non_ascii_chars = {
    char: count
    for char, count in character_counts.items()
    if ord(char) > 127
}

print("Total unique characters:", len(character_counts))
print("Non-ASCII unique characters:", len(non_ascii_chars))
```

```python

# Preserve original text
original_text = raw_text

# Create lowercase version
lowercase_text = raw_text.lower()

print("Original text sample:")
print(original_text[:50])

print("\nLowercase text sample:")
print(lowercase_text[:50])

# Compare character counts
print("\nOriginal characters:", len(original_text))
print("Lowercase characters:", len(lowercase_text))


```

```python
import re
# Normalize spaces and tabs within each line
lines = raw_text.splitlines()

cleaned_lines = [
    re.sub(r"[ \t]+", " ", line).strip()
    for line in lines
]


# Reconstruct text while preserving line breaks
whitespace_cleaned_text = "\n".join(cleaned_lines)

print("Original text sample:")
print(repr(raw_text[:300]))

print("\nWhitespace-cleaned text sample:")
print(repr(whitespace_cleaned_text[:300]))

print("\nOriginal character count:", len(raw_text))
print("Cleaned character count:", len(whitespace_cleaned_text))

print("\nOriginal preserved:", raw_text == original_text)
```

```python
# Remove Markdown heading symbols (##, ###, etc.)
markdown_cleaned_text = re.sub(
    r"(?m)^\s{0,3}#{1,6}\s+",
    "",
    whitespace_cleaned_text
)

print("Before Markdown cleaning:")
print(repr(whitespace_cleaned_text[:50]))

print("\nAfter Markdown cleaning:")
print(repr(markdown_cleaned_text[:50]))

print("\nOriginal character count:", len(whitespace_cleaned_text))
print("Markdown-cleaned character count:", len(markdown_cleaned_text))
```

```python
import string

# Count punctuation characters
punctuation_counts = Counter(
    char
    for char in markdown_cleaned_text
    if char in string.punctuation
)
print("Punctuation frequency:\n")

for punctuation, count in punctuation_counts.most_common():
    print(f"{repr(punctuation):8} : {count}")

print("\nTotal punctuation characters:", sum(punctuation_counts.values()))
print("Unique punctuation characters:", len(punctuation_counts))
```

```python
# Replace 3 or more consecutive newlines with 2 newlines
final_whitespace_text = re.sub(
    r"\n{3,}",
    "\n\n",
    markdown_cleaned_text
)

print("Before blank-line cleaning:")
print("Newline count:", markdown_cleaned_text.count("\n"))

print("\nAfter blank-line cleaning:")
print("Newline count:", final_whitespace_text.count("\n"))

print("\nOriginal text sample:")
print(repr(markdown_cleaned_text[:300]))

print("\nCleaned text sample:")
print(repr(final_whitespace_text[:300]))

print("\nRemaining excessive blank lines:",
      len(re.findall(r"\n{3,}", final_whitespace_text)))
```

```python
# Convert text to lowercase
final_cleaned_text = final_whitespace_text.lower()

# Final basic validation
print("Final preprocessing summary")
print("-" * 40)

print("Original characters:", len(raw_text))
print("Cleaned characters:", len(final_cleaned_text))

print("Original words:", len(raw_text.split()))
print("Cleaned words:", len(final_cleaned_text.split()))

print("Remaining Markdown headings:",
      len(re.findall(r"(?m)^\s{0,3}#{1,6}\s+", final_cleaned_text)))

print("Remaining excessive blank lines:",
      len(re.findall(r"\n{3,}", final_cleaned_text)))

print("\nFinal cleaned text sample:")
print(final_cleaned_text[:500])
```

```python
from pathlib import Path

# Create processed data directory
PROCESSED_DIR = PROJECT_DIR / "data" / "processed"
PROCESSED_DIR.mkdir(parents=True, exist_ok=True)

# Define output file
PROCESSED_DATA_PATH = PROCESSED_DIR / "cleaned_training_text.txt"

# Save cleaned text
with open(PROCESSED_DATA_PATH, "w", encoding="utf-8") as file:
    file.write(final_cleaned_text)

print("Processed dataset saved successfully!")
print("File path:", PROCESSED_DATA_PATH.resolve())
print("File exists:", PROCESSED_DATA_PATH.exists())
print("Saved characters:", PROCESSED_DATA_PATH.stat().st_size)
```
