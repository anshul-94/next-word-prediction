```python
from pathlib import Path
import re
import json
import numpy as np

# Project directory
PROJECT_DIR = Path.cwd().parent

# If your notebook is not running from notebooks/,
# use the actual project directory path instead.

DATA_PATH = PROJECT_DIR / "data" / "processed" / "cleaned_training_text.txt"

ARTIFACTS_DIR = PROJECT_DIR / "artifacts"

# Load cleaned dataset
with open(DATA_PATH, "r", encoding="utf-8") as file:
    cleaned_text = file.read()

# Load saved tokenizer
with open(ARTIFACTS_DIR / "word_to_index.json",
          "r", encoding="utf-8") as file:
    word_to_index = json.load(file)

# Convert vocabulary IDs to integers
word_to_index = {
    word: int(index)
    for word, index in word_to_index.items()
}

print("Notebook 04 initialized successfully!")
print("Dataset characters:", len(cleaned_text))
print("Vocabulary size:", len(word_to_index))
print("PAD ID:", word_to_index["<PAD>"])
print("OOV ID:", word_to_index["<OOV>"])
```

```python
# Split cleaned text into paragraphs
paragraphs = [
    paragraph.strip()
    for paragraph in re.split(r"\n\s*\n", cleaned_text)
    if paragraph.strip()
]

# Calculate paragraph statistics
paragraph_lengths = [
    len(paragraph.split())
    for paragraph in paragraphs
]

print("PARAGRAPH INSPECTION")
print("-" * 40)

print("Total paragraphs:", len(paragraphs))
print("Total words:", sum(paragraph_lengths))
print("Average words per paragraph:",
      round(np.mean(paragraph_lengths), 2))
print("Minimum words:", min(paragraph_lengths))
print("Maximum words:", max(paragraph_lengths))

print("\nFirst 3 paragraphs:")

for i, paragraph in enumerate(paragraphs[:3], start=1):
    print(f"\nParagraph {i}:")
    print(paragraph[:300])
    print("Word count:", len(paragraph.split()))
```
